```python
class CampusService:
    def __init__(self, service_name, location, fee):
        self.service_name = service_name
        self.location = location
        self.__fee = 0
        self.set_fee(fee)

    # Getter method
    def get_fee(self):
        return self.__fee

    # Setter method
    def set_fee(self, fee):
        if fee >= 0:
            self.__fee = fee
        else:
            print("Error: Fee cannot be negative.")

    # Base display method
    def display_service(self):
        print(f"Service Name: {self.service_name}")
        print(f"Location: {self.location}")
        print(f"Fee: KSh {self.get_fee()}")
```

```python
# First subclass
class LibraryService(CampusService):
    def __init__(self, service_name, location, fee, books_available):
        super().__init__(service_name, location, fee)
        self.books_available = books_available

    # Method overriding
    def display_service(self):
        print("\n--- Library Service ---")
        print(f"Service Name: {self.service_name}")
        print(f"Location: {self.location}")
        print(f"Fee: KSh {self.get_fee()}")
        print(f"Books Available: {self.books_available}")


# Second subclass
class TransportService(CampusService):
    def __init__(self, service_name, location, fee, vehicle_type):
        super().__init__(service_name, location, fee)
        self.vehicle_type = vehicle_type

    # Method overriding
    def display_service(self):
        print("\n--- Transport Service ---")
        print(f"Service Name: {self.service_name}")
        print(f"Location: {self.location}")
        print(f"Fee: KSh {self.get_fee()}")
        print(f"Vehicle Type: {self.vehicle_type}")
```

```python
# Create objects
library = LibraryService(
    "Digital Library",
    "Main Campus",
    200,
    5000
)

transport = TransportService(
    "Campus Shuttle",
    "Main Gate",
    50,
    "Electric Bus"
)

# Store objects in one list
campus_services = [library, transport]

# Demonstrate polymorphism
for service in campus_services:
    service.display_service()
```

```python
print("Current library fee:", library.get_fee())

library.set_fee(300)
print("Updated library fee:", library.get_fee())

library.set_fee(-100)
print("Library fee after invalid update:", library.get_fee())
```
